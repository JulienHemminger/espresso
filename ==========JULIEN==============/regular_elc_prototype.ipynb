```python
import espressomd
import espressomd.electrostatics

# 1. System Setup
# Note: ELC requires periodicity to be (True, True, True) 
# even though it simulates a 2D system.
box_l = 10.0
system = espressomd.System(box_l=[box_l, box_l, box_l])
system.time_step = 0.01
system.cell_system.skin = 0.4

# 2. Add Particles
# Particles must stay out of the "gap" region at the top of the box.
system.part.add(pos=[5.0, 5.0, 1.0], q=1.0)
system.part.add(pos=[5.0, 5.0, 7.0], q=-1.0)

# 3. Initialize the underlying 3D Solver (P3M)
# ELC acts as an extension/actor for P3M.
p3m = espressomd.electrostatics.P3M(
    prefactor=1.0, 
    accuracy=1e-3
)

# 4. Initialize ELC
# gap_size: The height of the empty region in z where no particles enter.
# maxPWerror: The maximum pairwise error allowed for the correction.
elc = espressomd.electrostatics.ELC(
    actor=p3m, 
    gap_size=2.0, 
    maxPWerror=1e-3
)

# 5. Activate the Solver
system.electrostatics.solver = elc

# Verify energy calculation
energy = system.analysis.energy()
print(f"Total Electrostatic Energy: {energy['total']}")
```
