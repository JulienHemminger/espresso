```python
# %%
import sys
import os
# To allow me to write/run juptyer notebooks from VScode.
espresso_build_path = "/home/main/Documents/Career/1_Studium/espresso/build"
sys.path.insert(0, os.path.join(espresso_build_path, "src", "python"))

import espressomd
import espressomd.electrostatics
import numpy as np

# 1. System Setup
BOX_L = 20.0
system = espressomd.System(box_l=[BOX_L, BOX_L, BOX_L])
system.time_step = 0.01
system.cell_system.skin = 0.4

# 2. Add Particles (The Dipole)
p1 = system.part.add(pos=[5.0, 5.0, 1.0], q=+1.0)
p2 = system.part.add(pos=[5.0, 5.0, 7.0], q=-1.0)

# 3. Setup Electrostatics (P3M)
# Prefactor = 1.0 (Bjerrum length = 1 in reduced units)
p3m = espressomd.electrostatics.P3M(prefactor=1.0, accuracy=1e-3)
system.electrostatics.solver = p3m

# 4. Calculate Energy and Forces
system.integrator.run(0) # Update system state
energy = system.analysis.energy()
force_p1 = p1.f
force_p2 = p2.f


"""
Total Energy: -0.18211 # TODO: ELC correction to remove unwanted z-interactions?
Force on Particle 0 (+): [0 0 0.0206293617]
Force on Particle 1 (-): [0 0 -0.0206293617]
"""

print(f"--- Dipole Analysis ---")
print(f"Total Energy: {energy['total']:.5f}")
print(f"Force on Particle 0 (+): {force_p1}")
print(f"Force on Particle 1 (-): {force_p2}")
```
